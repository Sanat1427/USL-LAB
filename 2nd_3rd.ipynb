```python
import math
from copy import deepcopy


# Given Data

labels = ["A", "B", "C", "D", "E"]
data = {
    "A": [10, 200, 0.5],
    "B": [12, 180, 0.6],
    "C": [20, 220, 0.4],
    "D": [25, 210, 0.9],
    "E": [22, 195, 0.7]
}


def euclidean(p1, p2):
    return math.sqrt(sum((a - b) ** 2 for a, b in zip(p1, p2)))


# (a) Distance Matrix

def distance_matrix(data):
    keys = list(data.keys())
    dist = {}
    for i in keys:
        dist[i] = {}
        for j in keys:
            dist[i][j] = euclidean(data[i], data[j])
    return dist

def print_matrix(dist):
    keys = list(dist.keys())
    print("     " + "  ".join(keys))
    for i in keys:
        row = [f"{dist[i][j]:6.2f}" for j in keys]
        print(f"{i}  " + " ".join(row))

dist_raw = distance_matrix(data)
print("\n(a) Distance Matrix (Without Normalization)")
print_matrix(dist_raw)

# Complete-Link Clustering

def complete_link_distance(cluster1, cluster2, dist):
    return max(dist[i][j] for i in cluster1 for j in cluster2)

def complete_link_clustering(dist):
    clusters = [[k] for k in dist.keys()]
    steps = []

    while len(clusters) > 1:
        min_dist = float("inf")
        pair = None

        for i in range(len(clusters)):
            for j in range(i + 1, len(clusters)):
                d = complete_link_distance(clusters[i], clusters[j], dist)
                if d < min_dist:
                    min_dist = d
                    pair = (i, j)

        i, j = pair
        merged = clusters[i] + clusters[j]
        steps.append((clusters[i], clusters[j], min_dist))

        clusters.pop(j)
        clusters.pop(i)
        clusters.append(merged)

    return steps

# (b) Complete-Link Clustering

steps_raw = complete_link_clustering(dist_raw)

print("\n(b) Complete-Link Clustering Steps (Without Normalization)")
for c1, c2, d in steps_raw:
    print(f"Merge {c1} and {c2} at distance {d:.2f}")

# (d) Normalize the Data (Z-score)

def normalize(data):
    keys = list(data.keys())
    cols = list(zip(*data.values()))
    means = [sum(col) / len(col) for col in cols]
    stds = [math.sqrt(sum((x - m) ** 2 for x in col) / len(col))
            for col, m in zip(cols, means)]

    norm = {}
    for k in keys:
        norm[k] = [(data[k][i] - means[i]) / stds[i] for i in range(len(means))]
    return norm

data_norm = normalize(data)

print("\n(d) Normalized Data")
for k in data_norm:
    print(k, ["{:.3f}".format(x) for x in data_norm[k]])


# (e) Distance Matrix

dist_norm = distance_matrix(data_norm)

print("\n(e) Distance Matrix (After Normalization)")
print_matrix(dist_norm)


# (f) Complete-Link Clustering

steps_norm = complete_link_clustering(dist_norm)

print("\n(f) Complete-Link Clustering Steps (After Normalization)")
for c1, c2, d in steps_norm:
    print(f"Merge {c1} and {c2} at distance {d:.2f}")
```
