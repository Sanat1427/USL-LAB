```python
import numpy as np
```

```python
labels=["A","B","C","D","E"]
#A>Data Matrix
data = np.array([
    [2, 1, 0, 3, 2, 1],
    [1, 2, 1, 2, 3, 0],
    [3, 0, 2, 1, 1, 2],
    [0, 3, 1, 3, 2, 2],
    [2, 2, 2, 0, 1, 3]
])

print("Data Matrix :")
for lbl,row in zip(labels,data):
    print(lbl,row)
```

```python
#manhattan distance
def manhattan_distance_matrix(X):
  n= X.shape[0]
  dist = np.zeros((n,n))
  for i in range(n):
    for j in range(n):
      dist[i,j]=np.sum(np.abs(X[i]-X[j]))
  return dist

manhattan_dist=manhattan_distance_matrix(data)

print(" Manhattan Distance Matrix:")
print("     " + "  ".join(labels))
for i, row in enumerate(manhattan_dist):
    print(f"{labels[i]}  " + "  ".join(f"{val:5.1f}" for val in row))
print()


```

```python
#c> euclidean distance
def euclidean_distance_matrix(X):
    n = X.shape[0]
    dist = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            dist[i, j] = np.sqrt(np.sum((X[i] - X[j]) ** 2))
    return dist

euclidean_dist = euclidean_distance_matrix(data)

print("c) Euclidean Distance Matrix:")
print("     " + "  ".join(labels))
for i, row in enumerate(euclidean_dist):
    print(f"{labels[i]}  " + "  ".join(f"{val:6.2f}" for val in row))
```
